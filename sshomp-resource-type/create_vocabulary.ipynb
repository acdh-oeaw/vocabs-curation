```python
voc = Graph()
concept_scheme = URIRef("https://vocabs.sshopencloud.eu/vocabularies/sshomp-resource-type/")
voc.add((concept_scheme, RDF.type, SKOS.ConceptScheme))
# title
voc.add((concept_scheme, SKOS.prefLabel, Literal("SSHOMP Resource Type", lang='en')))
voc.add((concept_scheme, DC.title, Literal("SSHOMP Resource Type", lang='en')))
# description
voc.add((concept_scheme, DC.description, Literal("""Controlled vocabulary of the resource types used to characterise items in the SSH Open Marketplace (https://marketplace.sshopencloud.eu). It supplies the values for the "resource type" property of Marketplace items, complementing the top-level item categories (tools & services, training materials, publications, datasets, workflows). The vocabulary is curated by the SSH Open Marketplace team and used both for manual curation and for automated ingest from partner sources.""", lang='en')))
# language
voc.add((concept_scheme, DC.language, Literal("English")))
# creators
voc.add((concept_scheme, DC.creator, Literal("Laure Barbot")))
voc.add((concept_scheme, DC.creator, Literal("Massimiliano Carloni")))
voc.add((concept_scheme, DC.creator, Literal("Vera Maria Charvát")))
voc.add((concept_scheme, DC.creator, Literal("Matej Ďurčo")))
voc.add((concept_scheme, DC.creator, Literal("Michael Kurzmeier")))
# contributors
voc.add((concept_scheme, DC.contributor, Literal("Laure Barbot")))
voc.add((concept_scheme, DC.contributor, Literal("Massimiliano Carloni")))
voc.add((concept_scheme, DC.contributor, Literal("Vera Maria Charvát")))
voc.add((concept_scheme, DC.contributor, Literal("Matej Ďurčo")))
voc.add((concept_scheme, DC.contributor, Literal("Michael Kurzmeier")))
voc.add((concept_scheme, DC.contributor, Literal("Klaus Illmayer")))
# publishers
voc.add((concept_scheme, DC.publisher, Literal("DARIAH-EU")))
voc.add((concept_scheme, DC.publisher, Literal("ACDH")))
# license
voc.add((concept_scheme, DCTERMS.license, Literal("https://creativecommons.org/publicdomain/zero/1.0/")))
# owner
voc.add((concept_scheme, DCTERMS.rightsHolder, Literal("DARIAH-EU")))
# subjects
voc.add((concept_scheme, DC.subject, Literal("social sciences and humanities", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("digital humanities", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("research infrastructures", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("research software and services", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("metadata", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("controlled vocabularies", lang='en')))
voc.add((concept_scheme, DC.subject, Literal("resource classification", lang='en')))
# relation
voc.add((concept_scheme, DC.relation, URIRef("https://marketplace.sshopencloud.eu")))
# coverage
voc.add((concept_scheme, DC.coverage, Literal("2019-", lang='en')))
# source
voc.add((concept_scheme, DC.source, Literal("Derived from the SSH Open Marketplace data model and its existing item-category and type values, consolidated and harmonised by the Marketplace curation team from the vocabularies of the ingested source collections, and from a comparison with 11 resource type lists or vocabularies (ATRIUM project semantic resources categories, EOSC resource categories, PARTHENOS entities, EOSC research resources from the federation handbook, FAIRsharing, COAR resource type, DCMI, ISO-19115-1, DataCite resource type, PARSE.Insight type, MARC Intellectual resource types)", lang='en')))
# version
voc.add((concept_scheme, OWL.versionInfo, Literal("1.0.0")))
```

### Load vocabulary data from CSV

```python
df = pd.read_csv('data.csv', skiprows=5, names=["top", "label_orig", "label_clean", "exact_match", "definition", "close_match", "comments", "uri"])
```

Replace all empty strings (or strings made just of spaces) with NaN, just to be sure

```python
df = df.replace(r'^\s*$', np.nan, regex=True)
```

```python
top_concepts = df.loc[df["top"].notna(), "label_clean"].tolist()
print(top_concepts)
```

```python
sub_concepts = df.loc[df["label_orig"].notna(), "label_clean"].tolist()
print(sub_concepts)
```

### Create vocabulary

```python
seen_uris = []

top_concept = ""

for _, row in df.iterrows():

    # Create URIs from the cleaned labels, lowering the case and replacing spaces with hyphens
    if pd.isna(row["uri"]):
        uri = row["label_clean"].lower().replace(" ","-").strip()
        uri = URIRef("https://vocabs.sshopencloud.eu/vocabularies/sshomp-resource-type/" + uri)
    else:
        uri = URIRef("https://vocabs.sshopencloud.eu/vocabularies/sshomp-resource-type/" + row["uri"])

    if uri in seen_uris:
        if pd.isna(row["top"]):
            voc.add((uri, SKOS.broader, top_concept))
        continue
    
    else:
        voc.add((uri, RDF.type, SKOS.Concept))
        seen_uris.append(uri)

        # prefLabel
        voc.add((uri, SKOS.prefLabel, Literal(row["label_clean"], lang='en')))

        # exactMatch
        if pd.notna(row["exact_match"]):
            voc.add((uri, SKOS.exactMatch, URIRef(row["exact_match"])))
            voc.add((uri, DC.source, URIRef(row["exact_match"])))

        # definition
        voc.add((uri, SKOS.definition, Literal(row["definition"], lang='en')))

        # # comments
        # if pd.notna(row["comments"]):
        #     voc.add((uri, SKOS.editorialNote, Literal(row["comments"], lang='en')))

        # closeMatches
        if pd.notna(row["close_match"]):
            for match in row["close_match"].split(";"):
                voc.add((uri, SKOS.closeMatch, URIRef(match.strip())))

        # declare it as topConcept or associate with broader concept
        if pd.notna(row["top"]):
            voc.add((uri, SKOS.topConceptOf, concept_scheme))
            top_concept = uri  
        else:
            voc.add((uri, SKOS.broader, top_concept))

voc.serialize(destination="sshomp.ttl")
```
