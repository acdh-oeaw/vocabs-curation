Load TTL into rdflib graph.

```python
from rdflib import Graph

g = Graph()
g.parse("terralid_skosified.ttl", format="turtle")

print(f"Loaded {len(g)} triples.")
```

Find concept schemes referenced by property `skos:inScheme`.

```python
query = """
SELECT DISTINCT ?scheme
WHERE {
    ?concept a skos:Concept.
    ?concept skos:inScheme ?scheme.
}"""

response = g.query(query)
print(len(response), "concept schemes found.")
for row in response:
    print(row)
```

Find number of concepts.

```python
query = """
SELECT DISTINCT ?concept
WHERE {
    ?concept a skos:Concept.
}"""

response = g.query(query)
print(len(response))
```

Find distinct properties.

```python
query = """
SELECT DISTINCT ?p
WHERE {
    ?s ?p ?o.
}"""

response = g.query(query)
print(len(response))
for row in response:
    print(row)
```
